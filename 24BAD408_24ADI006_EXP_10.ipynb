```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import torch

model_name = "google/flan-t5-base"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

device = "cuda" if torch.cuda.is_available() else "cpu"
model = model.to(device)

print("Model loaded successfully!")
```

```python
# Name: SANTHOSH PRABHU S
# Roll No: 24BAD408
import torch
knowledge_base = {
    "what is artificial intelligence":
        "Artificial Intelligence (AI) is a branch of computer science that enables machines to perform tasks that usually require human intelligence. These tasks include learning, reasoning, problem-solving, and understanding language. For example, voice assistants use AI to understand spoken commands. AI is used in healthcare, education, banking, and transportation. Its goal is to develop systems that can perform intelligent tasks.",

    "what is ai":
        "Artificial Intelligence (AI) enables machines to perform tasks that usually require human intelligence, such as learning, reasoning, and problem-solving. AI is used in healthcare, education, banking, and transportation.",

    "what is machine learning":
        "Machine Learning (ML) is a branch of Artificial Intelligence that allows computers to learn patterns from data. Instead of programming every rule manually, developers train models using examples. The model uses learned patterns to make predictions or decisions. For example, an email spam filter learns to identify unwanted emails. ML is used in recommendation systems, fraud detection, and image recognition.",

    "what is deep learning":
        "Deep Learning is a subfield of machine learning that uses neural networks with multiple layers. These networks learn complex patterns from large amounts of data. Deep learning is useful for processing images, speech, and text. It is used in autonomous vehicles, medical imaging, and language models.",

    "what is natural language processing":
        "Natural Language Processing (NLP) is a field of AI that helps computers process and understand human language. NLP techniques include text classification, translation, and sentiment analysis. It is used in chatbots, voice assistants, and search engines.",

    "what is nlp":
        "Natural Language Processing (NLP) helps computers understand and process human language. It is used in translation, sentiment analysis, chatbots, and voice assistants.",

    "what is a chatbot":
        "A chatbot is a software application designed to communicate with users through text or speech. It can answer questions, provide information, and help users complete tasks. Chatbots are used in education, customer support, banking, and healthcare.",

    "what is a large language model":
        "A Large Language Model (LLM) is an AI model trained on large amounts of text to learn language patterns. It can generate text, answer questions, summarize information, and assist with writing. LLMs commonly use transformer-based neural networks.",

    "what is llm":
        "A Large Language Model (LLM) is an AI model trained on large amounts of text. It can answer questions, generate content, summarize information, and assist with writing.",

    "what is generative ai":
        "Generative AI is a type of artificial intelligence that creates new content based on patterns learned during training. It can generate text, images, audio, video, and computer code. It is used in education, design, and software development.",

    "what is a neural network":
        "A neural network is a machine learning model inspired by the connected structure of biological neurons. It consists of layers of interconnected artificial neurons that process information. Neural networks are used in speech recognition, computer vision, and deep learning.",

    "what is supervised learning":
        "Supervised learning is a machine learning approach that uses labelled training data. The model learns the relationship between inputs and expected outputs to make predictions for new data. Classification and regression are common supervised learning tasks.",

    "what is unsupervised learning":
        "Unsupervised learning is a machine learning approach that works with data without predefined labels. It discovers patterns, structures, or groups in data. Clustering is a common technique used for customer segmentation and data exploration.",

    "what is reinforcement learning":
        "Reinforcement learning is a machine learning approach in which an agent learns by interacting with an environment. It receives rewards or penalties based on its actions and learns a strategy to maximize rewards. It is used in robotics and games.",

    "what is gradio":
        "Gradio is a Python library for creating user interfaces for machine learning models. It connects Python functions to interactive components such as text boxes and buttons. It helps developers build and share machine learning applications.",

    "what is hugging face transformers":
        "Hugging Face Transformers is a Python library that provides tools for using pretrained machine learning models. It supports text generation, translation, classification, and question answering.",

    "what is a pretrained model":
        "A pretrained model is a machine learning model that has already been trained on a dataset. Developers can use it directly or fine-tune it for a specific task. Pretrained models can reduce development time and computational requirements.",

    "what is prompt engineering":
        "Prompt engineering is the process of designing instructions or questions to guide an AI model toward a useful response. Clear prompts can help models produce more relevant answers.",

    "what is a dataset":
        "A dataset is a collection of related information used for analysis or machine learning. It may contain text, images, numerical values, or other data. The quality of a dataset can affect model performance.",

    "what is the difference between ai and machine learning":
        "Artificial Intelligence is the broader field of building systems that perform tasks requiring intelligent behaviour. Machine Learning is a subset of AI that enables systems to learn patterns from data. Therefore, all machine learning is part of AI, but not all AI relies on machine learning.",

    "what are the applications of ai":
        "Artificial Intelligence has applications in healthcare, education, banking, agriculture, transportation, manufacturing, and customer service. AI can help analyse medical images, detect suspicious transactions, and support personalised learning."
}


greetings = {
    "hi": "Hi! How can I help you today?",
    "hello": "Hello! Welcome to EduBot AI - Your Learning Assistant 🤖. What would you like to know?",
    "hey": "Hey! How can I help you?",
    "good morning": "Good morning! Have a great day. How can I help you?",
    "good afternoon": "Good afternoon! How can I assist you?",
    "good evening": "Good evening! What would you like to learn today?",
    "how are you": "I'm doing great! Thanks for asking. How can I help you?",
    "thank you": "You're welcome! Happy to help.",
    "thanks": "You're welcome! Feel free to ask me anything.",
    "bye": "Goodbye! Have a wonderful day!",
    "goodbye": "See you again! Take care.",
    "good night": "Good night! Sleep well.",
    "who are you": "I'm EduBot AI - Your Learning Assistant 🤖. I'm here to answer your questions.",
    "what is your name": "I'm EduBot AI - Your Learning Assistant 🤖!",
    "what can you do": "I can answer questions about AI, machine learning, science, technology, programming, and general knowledge.",
    "help": "Ask me anything about AI, technology, science, programming, or general knowledge."
}


def generate_response(question):

    if not question or not question.strip():
        return "Please enter a message."

    key = " ".join(
        question.lower().strip().rstrip("?.!,").split()
    )

    # Handle greetings and common messages
    if key in greetings:
        return greetings[key]

    # Handle common greeting phrases
    if key.startswith("hi ") or key.startswith("hello ") or key.startswith("hey "):
        return greetings[key.split()[0]]

    # Handle existing AI knowledge-base questions
    if key in knowledge_base:
        return knowledge_base[key]

    # Handle common variations of AI questions
    if "artificial intelligence" in key or key in ["explain ai", "explain artificial intelligence"]:
        return knowledge_base["what is artificial intelligence"]
    if "machine learning" in key and key.startswith(("what is", "explain", "define")):
        return knowledge_base["what is machine learning"]
    if "deep learning" in key and key.startswith(("what is", "explain", "define")):
        return knowledge_base["what is deep learning"]
    # Use the pretrained model for other general and technical questions
    prompt = (
        "You are a friendly and helpful AI chatbot. "
        "Answer general knowledge, everyday, science, technology, "
        "and programming questions in simple English. "
        "Give a clear and relevant answer with an example when useful. "
        "Do not invent facts if you are unsure.\n"
        "Question: " + question + "\n"
        "Answer:"
    )

    try:
        inputs = tokenizer(
            prompt,
            return_tensors="pt",
            truncation=True,
            max_length=512
        ).to(device)

        with torch.no_grad():
            outputs = model.generate(
                **inputs,
                max_new_tokens=150,
                num_beams=4,
                do_sample=False
            )

        answer = tokenizer.decode(
            outputs[0],
            skip_special_tokens=True
        ).strip()

        return answer if answer else "Sorry, I couldn't generate an answer. Please try again."

    except Exception as e:
        return "Sorry, something went wrong: " + str(e)


# Test the chatbot
questions = [
    "Hi",
    "Hello",
    "How are you?",
    "What is your name?",
    "What is Artificial Intelligence?",
    "What is AI?",
    "What is machine learning?",
    "Explain deep learning",
    "What is NLP?",
    "What is Gradio?",
    "What is the capital of India?",
    "What is Python?",
    "Thank you",
    "Bye"
]

for question in questions:
    print("Question:", question)
    print("Answer:", generate_response(question))
    print("-" * 60)
```

```python
import gradio as gr

demo = gr.Interface(
    fn=generate_response,

    inputs=gr.Textbox(
        lines=3,
        label="Ask EduBot AI",
        placeholder="Type Hi, ask a general question, or learn about AI..."
    ),

    outputs=gr.Textbox(
        lines=8,
        label="EduBot AI Response"
    ),

    title="EduBot AI - Your Learning Assistant 🤖",

    description=(
        "Welcome to EduBot AI! 👋 "
        "Chat with me using greetings like Hi or Hello, "
        "ask general knowledge questions, or learn about "
        "Artificial Intelligence, Machine Learning, Deep Learning, "
        "NLP, and other technical topics."
    ),

    examples=[
        "Hi",
        "Hello",
        "How are you?",
        "What is your name?",
        "What can you do?",
        "What is Artificial Intelligence?",
        "What is machine learning?",
        "Explain deep learning.",
        "What is a chatbot?",
        "What is prompt engineering?",
        "What is the capital of India?",
        "Explain Python programming.",
        "Thank you",
        "Bye"
    ],

    submit_btn="Send Message",
    clear_btn="Clear"
)

demo.launch(share=True)
```
